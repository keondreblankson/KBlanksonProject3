For this project, I used the Chinook database to show each customer's last name, first name, purchased tracks, and album titles in a pandas DataFrame. I sorted the results by last name and then first name.


```python
import sqlite3
from pathlib import Path
from urllib.request import urlretrieve
import pandas as pd

# Download the database if it is not already in the folder.
db_path = Path("Chinook_Sqlite.sqlite")
db_url = (
    "https://raw.githubusercontent.com/lerocha/chinook-database/"
    "master/ChinookDatabase/DataSources/Chinook_Sqlite.sqlite"
)
if not db_path.exists():
    urlretrieve(db_url, db_path)
```

### Joining the tables
I joined Customer, Invoice, InvoiceLine, Track, and Album using their matching IDs. These tables connect the customers to their purchases and show which album each track belongs to.


```python
query = """
SELECT
    c.LastName,
    c.FirstName,
    t.Name,
    a.Title
FROM Customer AS c
INNER JOIN Invoice AS i
    ON c.CustomerId = i.CustomerId
INNER JOIN InvoiceLine AS il
    ON i.InvoiceId = il.InvoiceId
INNER JOIN Track AS t
    ON il.TrackId = t.TrackId
INNER JOIN Album AS a
    ON t.AlbumId = a.AlbumId
ORDER BY c.LastName, c.FirstName, il.InvoiceLineId;
"""

with sqlite3.connect(db_path) as connection:
    purchases = pd.read_sql_query(query, connection)

purchases.head()
```

```python
print("Total purchase rows:", format(len(purchases), ","))
print("Columns:", ", ".join(purchases.columns))
```

### Results
The DataFrame has 2,240 rows. Each row shows a purchased track along with the customer's name and the album title. The results are sorted by last name and first name. I used head() to display the first five rows.
